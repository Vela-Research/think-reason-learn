# VerifiableRL
Learns which slot of information to reveal next, at a cost (Ye et al., 2026). Needs `pip install 'think-reason-learn[rl]'`.

```python
import numpy as np
from think_reason_learn.verifiable_rl import VerifiableRL

rng = np.random.default_rng(0)
slots = {"education": 8, "jobs": 8, "funding": 8}  # columns per slot, in order
X = rng.normal(size=(200, 24)).astype(np.float32)
# The "jobs" slot carries the signal.
y = np.where(X[:, 8] + X[:, 9] > 0, "hit", "miss")
```

```python
model = VerifiableRL(
    slots,
    slot_costs={"funding": 0.5},
    policy_hidden=64,
    clf_hidden=64,
    n_iterations=10,
    pos_label="hit",
    random_state=0,
)
model.fit(X, y)
(model.predict(X) == y).mean()
```
