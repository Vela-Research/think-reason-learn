# Policy Induction
A policy refined case by case, each refinement kept only when it raises precision (Mu et al., 2025).

Set the API key of the provider you use (for example `OPENAI_API_KEY`) and, for Jev, `TYPESAFE_API_KEY`. Any reasoning model can answer instead of Jev: pass `answerer="openai:gpt-5-mini"`.

```python
X = [
    "Alex is a serial entrepreneur with two exits and a strong network in AI.",
    "Jordan graduated top of class from MIT but has no business experience.",
    "Taylor has 10 years in finance and raised a seed round quickly.",
    "Casey started a company out of high school and failed twice.",
    "Morgan is a former Google engineer with patents in machine learning.",
    "Riley left a consulting job last month and has no product yet.",
    "Sam is a staff engineer who sold a company to Stripe.",
    "Drew is a first-time founder with a marketing background.",
]
y = ["successful", "failed"] * 4
```

```python
from think_reason_learn.policy_induction import PolicyInduction

pi = PolicyInduction(
    task="Predict whether a startup founder will succeed.",
    generator="openai:gpt-5-mini",
    answerer="jev",
    n_samples=8,  # the paper refines with 240 cases over 4 rounds
    n_rounds=1,
    batch_size=2,
    pos_label="successful",
    random_state=0,
)
pi.fit(X, y)
print("\n".join(pi.policy_))
```

```python
pi.history_
```

```python
pi.predict_proba(["Jamie is a robotics PhD who scaled a team to 40 engineers."])
```
